```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Flatten

# Cargar dataset
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# Normalizar datos
x_train = x_train.astype('float32') / 255.
x_test = x_test.astype('float32') / 255.

# Aplanar imágenes de 28x28 a 784
x_train_flat = x_train.reshape((len(x_train), -1))
x_test_flat = x_test.reshape((len(x_test), -1))

# One-hot encoding
y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)
```

```python
# Arquitectura del Autoencoder
input_img = Input(shape=(784,))
encoded = Dense(512, activation='relu')(input_img)
encoded = Dense(256, activation='relu')(encoded)
encoded = Dense(64, activation='relu')(encoded)

decoded = Dense(256, activation='relu')(encoded)
decoded = Dense(512, activation='relu')(decoded)
decoded = Dense(784, activation='sigmoid')(decoded)

# Modelo Autoencoder
autoencoder = Model(input_img, decoded)

# Entrenar el autoencoder
autoencoder.compile(optimizer='adam', loss='binary_crossentropy')
autoencoder.fit(x_train_flat, x_train_flat,
                epochs=20,
                batch_size=256,
                shuffle=True,
                validation_data=(x_test_flat, x_test_flat))
```

```python
# Crear el modelo del codificador
encoder = Model(input_img, encoded)

# Obtener las características codificadas
x_train_encoded = encoder.predict(x_train_flat)
x_test_encoded = encoder.predict(x_test_flat)
```

```python
# Construir un clasificador simple usando las características codificadas
input_encoded = Input(shape=(64,))
classifier_output = Dense(32, activation='relu')(input_encoded)
classifier_output = Dense(10, activation='softmax')(classifier_output)

classifier = Model(input_encoded, classifier_output)

# Compilar y entrenar
classifier.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
classifier.fit(x_train_encoded, y_train_cat,
               epochs=20,
               batch_size=128,
               validation_data=(x_test_encoded, y_test_cat))
```

```python
loss, accuracy = classifier.evaluate(x_test_encoded, y_test_cat)
print(f"Precisión del clasificador sobre codificaciones: {accuracy * 100:.2f}%")
```
